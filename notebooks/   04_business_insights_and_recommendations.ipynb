## 1. International customers are a smaller but higher-value segment

**Finding:** Among customers with a Customer ID, the UK generates 33,389 orders and £14,294,286.13 in revenue, while international markets generate 3,261 orders and £2,791,338.18. International accounts for 8.9% of these orders but 16.3% of this revenue, because its average order value (£855.98) is roughly double the UK's (£428.11).

**Why it matters:** On average, each international order contributes about twice the revenue of a UK order (a typical order about 1.4×), so international is a higher-value segment that is currently small.. Growth there could add more revenue per order won than the same growth in the UK. This is revenue only; the dataset has no cost or margin data, so profitability can't be assessed.

**Recommendation:** First, protect and grow the top international accounts, since 5 customers generate 46.71% of international revenue (for example, a named account contact and priority service). Second, run a small test in one or two of the strongest international markets, such as targeted marketing, and measure whether order count grows while the typical order value holds. Scale the broader push only if the test works.

**Caveat:** The "double the UK" gap is partly driven by a few large orders. On the median, a typical international order (£419.85) is about 1.4× a typical UK order (£299.41), not 2×. In addition, 5 of 526 international customers (under 1%) generate £1,303,708.33, or 46.71% of international revenue, so the segment depends heavily on a handful of accounts. Guest orders (no Customer ID, 13.3% of total revenue) can't be assigned to a country and are excluded.

## 2. Revenue depends on a small group of repeat customers

**Finding:** Repeat customers (4,236, or 72.23% of identified customers) generate 96.74% of identified-customer revenue (£16,528,749.89), while 1,629 one-time buyers generate 3.26%. (The top 20% of customers (1,173) generate 77.20% of that revenue. In the RFM segmentation, 1,831 "Champions" generate 76.23% of it (£13,024,092.78).

**Why it matters:** Revenue is driven by a relatively small group of loyal, high-spending buyers rather than by attracting many one-time customers. Losing even a few Champions would have an outsized effect on revenue. At the other end, 1,799 "Lost" contribute only 3.35% of identified-customer revenue, which suggests a large dormant group.

**Recommendation:** Make retention the priority. Give Champions and Loyal customers priority service or early access to new lines, and flag any Champion whose gap between orders is growing (the "At Risk" segment, 439 customers, is the early warning). Run a low-cost win-back campaign for the Lost group and track how many place a new order.

**Caveat:** Customer-level figures only cover transactions with a Customer ID; about 22.8% of rows have none, so guest purchases are excluded from this analysis. RFM segment boundaries come from quartile scoring, which is a modelling choice, and the data covers only two years, so long-term loyalty can't be assessed.

**Note on revenue figures:** **Caveat:** Customer-level figures only cover transactions with a Customer ID. About 22.8% of rows have none, and these guest orders are 13.3% of total revenue (£2,614,956.24), so they are excluded. RFM segment boundaries come from quartile scoring, which is a modelling choice. Some "one-time" buyers may be recent customers whose second order falls after the data ends in December 2011, and two years of data can't show long-term loyalty.

## 3. Revenue is strongly seasonal, peaking in November

**Finding:** Monthly revenue builds through autumn and peaks in November in both years. In 2010 it rose from £869,727.16 in September to £1,095,413.63 in October and £1,437,823.55 in November, then fell to £789,858.83 in December.

**Why it matters:** The pattern is consistent with a wholesale gift retailer whose business customers stock up before Christmas. Stock, staffing and cash flow need to be in place before the peak, not during it, and the December drop is a normal pattern rather than a warning sign.

**Recommendation:** Plan stock and fulfilment capacity ahead of September–November. Contact trade customers in August or September with pre-order or early-booking offers to capture peak demand and spread the workload. Consider a smaller promotion in the quieter months (January–March) to lift the low points.

**Caveat:** The data contains only two Novembers, so the pattern is suggestive, not proven. December 2011 holds only 9 days of data, so that year's drop is overstated. The data shows revenue, not stock levels or capacity, so it can't show whether the November peak caused any fulfilment problems.

## 4. Cancellations are steady overall, but a few products stand out

**Finding:** Cancelled transactions are 1.85% of all transactions (19,104 of 1,033,031). The monthly cancellation rate is stable at roughly 1.3%–2.8% with no seasonal spike; November 2011, the peak revenue month, had one of the lowest rates (1.29%). Among real products, REGENCY CAKESTAND 3 TIER was cancelled most often (341 cancellations, £16,545.30), followed by WHITE HANGING HEART T-LIGHT HOLDER (134 cancellations, £9,387.10).

**Why it matters:** The steady rate suggests fulfilment held up during the busiest period. The two most-cancelled products may point to a product-specific issue (quality, listing accuracy or damage in transit), and that is cheaper to investigate than a general cancellation problem.

**Recommendation:** Calculate the cancellation rate per product (cancellations divided by units sold) for the top sellers, then investigate the highest ones: check listing descriptions, packaging and customer feedback. Track the rate for those products over the following months.

**Caveat:** A high cancellation count isn't the same as a high rate. Regency Cakestand is also a top seller, so its count may partly reflect volume. The dataset has no cancellation reasons, so any cause is a hypothesis. A few large cancelled values (for example one £168,469.60 cancellation of PAPER CRAFT, LITTLE BIRDIE) reflect single big orders, not a pattern.
